**Normalización** Convertir los datos de entrada en un rango de 0 a 1, a este proceso le llamamos homotencia.

**Formula:**
$x_{nuevo} = \frac{x - x_{min}}{x_{max} - x_{min}}$

```python
from sklearn import preprocessing
import numpy as np
```

**MinMaxScaler:** Función que permite encontar los valores minimos y maximos de un conjunto de datos

**Numpy:** Una funcion de python desarrollada en C que me permite realizar operaciones matematicas complejas con estrcutruras de datos como matrices, vectores.

```python
# Carga de datos.
X_train = np.array([[1., -4., 5.],
                    [5., 0., 2.],
                    [0., 6., -3.]])

print(X_train)
```

```python
# Limpieza de datos: normalización.
normalizer = preprocessing.MinMaxScaler()
X_train_norm = normalizer.fit_transform(X_train)
# Salida de resultados.
print(X_train_norm)
```

**Fit**
Sirve para analizar los datos y calcular los valores necesarios para poder normalizarlos posteriormente.

**Transform**
Sirve para transformar los datos originales utilizando los valores que calculó previamente

```python
# Ejemplo para ver que los resultados son iguales si aplicamos los métodos
# "fit" y "transform" por separado

X_train_2 = np.copy(X_train)

# Crea un objeto "fitted_normalizer" llamando al método "fit"
f = normalizer.fit(X_train)

# Aplica el objeto sobre los datos "X_train_2"
# para obtener "X_train_norm_2"
X_train_norm_2 = f.transform(X_train)

# Comprueba si "X_train_norm" y "X_train_norm_2" son iguales
print(np.array_equiv(X_train_norm, X_train_norm_2))
```

```python
# Normalizar un nuevo conjunto de datos de test
# (SOLO HAY QUE APLICAR EL MÉTODO "transform", NO EL "fit")

X_test = np.array([[1., -3., 2.],
                   [3., 0., 0.],
                   [0., 4., -1.]])

# Normalizar los datos de "X_test"
X_test_norm = f.transform(X_test)
print(X_test_norm)
```
